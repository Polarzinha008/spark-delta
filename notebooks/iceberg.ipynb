```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("IcebergExample") \
    .config("spark.jars.packages", "org.apache.iceberg:iceberg-spark-runtime-3.5_2.12:1.4.2") \
    .config("spark.sql.catalog.spark_catalog", "org.apache.iceberg.spark.SparkSessionCatalog") \
    .config("spark.sql.catalog.spark_catalog.type", "hadoop") \
    .config("spark.sql.catalog.spark_catalog.warehouse", "/tmp/iceberg-warehouse") \
    .getOrCreate()
```

```python
spark.sql("SHOW DATABASES").show()
```

```python
spark.sql("CREATE DATABASE db_vendas")
```

```python
spark.sql("""
CREATE TABLE spark_catalog.db_vendas.vendas (
    id INT,
    produto STRING,
    preco INT
) USING iceberg
""")
```

```python
spark.sql("""
INSERT INTO spark_catalog.db_vendas.vendas VALUES
(1, 'Notebook', 3000),
(2, 'Mouse', 100),
(3, 'Teclado', 200)
""")
```

```python
spark.sql("SELECT * FROM spark_catalog.db_vendas.vendas").show()
```

```python
spark.sql("""
UPDATE spark_catalog.db_vendas.vendas
SET preco = 150
WHERE id = 2
""")
```

```python
spark.sql("SELECT * FROM spark_catalog.db_vendas.vendas").show()
```

```python
spark.sql("""
DELETE FROM spark_catalog.db_vendas.vendas
WHERE id = 3
""")
```
